```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/ACADEMIC-AND-PERSONAL-PROJECTS/Homework-1-Introduction-to-Machine-Learning/main/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)

print("Q1 pandas:", pd.__version__)
print("Q2 registros:", len(df))
print("Q3 tipos de combustível:", df["fuel_type"].nunique())
print("Q4 colunas com missing:", (df.isnull().sum() > 0).sum())
print("Q5 max eficiência (Asia):", df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max())

antes = df["horsepower"].median()
moda = df["horsepower"].mode()[0]
df["horsepower"] = df["horsepower"].fillna(moda)
depois = df["horsepower"].median()
print("Q6 mediana antes/depois:", antes, depois)

asia = df[df["origin"] == "Asia"]
X = asia[["vehicle_weight", "model_year"]].head(7).to_numpy()
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print("Q7 soma dos pesos:", w.sum())
```
